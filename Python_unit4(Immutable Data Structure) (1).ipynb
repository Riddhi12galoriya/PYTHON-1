```python
s="FSfds@#4fsf323"
alphabets=0
upper=0
lower=0
digit=0
special_symbols=0
for i in s:
    if i.isalpha()==True:
        alphabets+=1
        if i.isupper()==True:
            upper+=1
        else:
            lower+=1
    elif i.isdigit()==True:
        digit+=1
    elif i.isspace():
        pass
    else:
        special_symbols+=1
print("Alphabets:",alphabets)
print("Upper:",upper)
print("Lower:",lower)
print("Digits:",digit)
print("Special Symbols:",special_symbols)
```

```python
#to find sum and average of upper program
s = "FSfds@#4fsf323"
sum = 0
alphabets = 0
upper = 0
lower = 0
digit = 0
special_symbols = 0
for i in s:
    if i.isalpha() == True:
        alphabets+=1
        if i.isupper() == True:
            upper+=1
        else:
            lower+=1
    elif i.isdigit() == True:
        digit+=1
        sum+=1
    elif i.isspace():
        pass
    else:
        special_symbols+=1
average = sum / digit
print("Alphabets:",alphabets)
print("Upper:",upper)
print("Lower:",lower)
print("Digits:",digit)
print("Special Symbols:",special_symbols)
print("Sum:",sum)
print("Average:",average)
```
