# Task - 1 
WAF for each of following string based tasks:

- Vowel and Consonent Counter
- Word level Sentence Reversal
- Vowel Masking(Replacement)

```python
def v_consonant(t):
    vowels = "aeiouAEIOU"
    v_count = 0
    c_count = 0
    for i in t:
        if i.isalpha():
            if i in vowels:
                v_count+=1
            else:
                c_count+=1
    return v_count,c_count
a = str(input("enter string:"))
v, c = v_consonant(a)
print("Vowels:",v)
print("Consonant:",c)
```

```python
#Word level sentence reversal
str1 = input("Enter String:")
def reverse_word(s):
    words = s.split()
    reversed_list = words[::-1]
    return " ".join(reversed_list)
r = reverse_word(str1)
print(r)
```

```python
#Vowel Masking(Replacement)
str1 =input("Enter String:")
def mask_vowels(s):
    vowels = "aeiouAEIOU"
    masked_string = " "

    for char in s:
        if char in vowels:
            masked_string+="*"
        else:
            masked_string+=char
    return masked_string
m = mask_vowels(str1)
print(m)
```

# Task - 2
WAF to solve the following:

- Acronym Extraction: split into two words, extract first letter of each word, convert them to uppercase, join them together to form base acronym string.
- Digit Sum Injection: scan entire original string for any numeric digits, compute the sum of all digits found, append this sum directly to the end of acronym. If no digits present in string, append 0.
- Vowel Parity Suffix: Count total number of vowels(uppercase and lowercase) in original string. If total count= even append even. If total count= odd append odd. Return the final combined string.

```python
def generate_code(s):
    #1. Acronym Extraction
    words = s.split()
    acronym = ""
    for i in words:
        if i:
            acronym += i[0].upper()
    
    #2. Digit Sum Injection
    digit_sum = 0
    for char in s:
        if char.isdigit():
            digit_sum += int(char)
    code = acronym + str(digit_sum)

    #3. Vowel Parity Suffix
    vowels = "aeiouAEIOU"
    v_count = 0
    for i in s:
        if i in vowels:
            v_count +=1
    print(v_count)

    if v_count%2==0:
        final_code = code + "-EVEN"
    else:
        final_code = code + "_ODD"
    return final_code

print(generate_code("python programming"))
print(generate_code("Data Science 404"))
```

```python
def calculate_pizza_price(size, toppings, cheese):
    cost = 0
    # Calculate pizza base price based on size
    if size == 'small':
        cost += 50
    elif size == 'medium':
        cost += 100
    else:
        cost += 200
    
    # Calculate topping prices
    topping_prices_20 = ['corn', 'tomato', 'onion', 'capsicum']
    topping_prices_50 = ['mushroom', 'olives', 'broccoli']
 
    for topping in toppings:
         cost += 20 if topping in topping_prices_20 else 50
    #Calculate cheese price
    cost += 50 * len(cheese)
    return cost

def order_pizza():
    pizzas = []
    number = int(input("How many pizzas you want to order: "))
    for i in range(number):
        toppings = []
        cheese = []
        print('Customize Pizza', i + 1)
        print('Select Pizza Size: small, medium or large')
        size = input('Select size: ')
        
        print('''Select Toppings: 
                20 for 'corn', 'tomato', 'onion', 'capsicum'
                50 for 'mushroom', 'olives', 'broccoli' ''')
        toppings_count = int(input('How many toppings: '))
    
        for _ in range(toppings_count):
             toppings.append(input('Enter toppings: '))
        
        print('Select Cheese: mozarrella, feta, cheddar')
        cheese_count = int(input('How many cheese: '))

        for _ in range(cheese_count):
             cheese.append(input('Enter cheese: '))
        print()
        pizza_cost = calculate_pizza_price(size, toppings, cheese)
        pizzas.append((size, toppings, cheese, pizza_cost))
    return pizzas

def print_bill(pizzas):
     total = 0
     count = 1
     for pizza in pizzas:
         print('Pizza', count)
         print("Size:", pizza[0])
         print("Toppings:", pizza[1])
         print("Cheese:", pizza[2])
         print("Pizza cost:", pizza[3])
         total += pizza[3]
         count += 1
     print('Total bill amount:', total)
# Main program
pizza_order = order_pizza()
print_bill(pizza_order)
```

# Task - 4
Inventory Analysis System

inventory = {
"Laptop": {"price": 55000, "quantity": 8, "category": "Electronics"},
"Mobile": {"price": 25000, "quantity": 15, "category": "Electronics"},
"Chair": {"price": 3500, "quantity": 20, "category": "Furniture"},
"Table": {"price": 7000, "quantity": 10, "category": "Furniture"}
}

1. Calculate the total inventory value of every product.
2. Create a dictionary containing product : total_value.
3. Find products whose total inventory value exceeds ₹2,00,000.
4. Find all unique categories using a set.
5. Find the product having maximum inventory value using lambda.
6. Sort products according to inventory value in descending order.
7. Use dictionary comprehension wherever appropriate.

```python
inventory = {
"Laptop":{"price":55000, "quantity":8, "category":"Electronics"},
"Mobile":{"price":25000, "quantity":15, "category":"Electronics"},
"Chair":{"price":3500, "quantity":20, "category":"Furniture"},
"Table":{"price":7000, "quantity":10, "category":"Furniture"}
}
```

```python
# 1 & 2 Total inventory value of each product -> {product:total_value}
total_value = {item: d["price"] *d["quantity"] for item, d in inventory.items()}
print("1-2) Total value per product:",total_value)
```

```python
#3) Product whose total value exceeds Rs 2,00,000
high_value = {item: v for item, v in total_value.items() if v>200000}
print("3) Value > 2,00,000:",high_value)
```

```python
#4) Unique categories using a set
categories = {d["category"] for d in inventory.values()}
print("4) Unique Categories are:",categories)
```

```python
#5) Product with maximum inventory value using lambda:
max_item = max(total_value.items(),key=lambda kv: kv[1])
print(max_item)
print("5) Max inventory value:",max_item[0],"->Rs",f"{max_item[1]:,}")
```

```python
#6) Sort products by inventory value
sorted_items = sorted(total_value.items(), key = lambda kv: kv[1], reverse=True)
print("6) Sorted (Descending):")
for name, val in sorted_items:
    print(f"{name:} Rs {val:}")
```

```python
#7) Extra: dict comprehension -> category-wise total value
category_totals = {c:sum(v for p, v in total_value.items() if inventory[p]["category"] == c)
                  for c in categories}
print("7) Category-wise Totals:",category_totals)
```
