# Accessing Characters of String:
**1) Using index**

**2) Using Slicing Operators**

```python
# Using Index operator
s="Advait"
print(s[3])
print(s[-3])
```

```python
# Using Slicing Operators
s="Learning Python is very easy. "
print(s[1:7:1])
print(s[1:7:2])
print(s[1:7])
print(s[:7])
print(s[:9:])
print(s[:])
print(s[::])
print(s[::-1])
print(s[-4:])
print(s[-4:-1]) # this gives empty string
print(s[-4::-1])
```

# Important Functions of String:


```python
# len function:
s="LJ UNIVERSITY"
print(len(s))
```

```python
# removing spaces of string includes: 
#(I)lstrip(): for left side space removal 
#(II)rstrip(): for right side space removal
#(III)strip(): for all space removal
s="   Python   "
x=s.lstrip()
print(x)
y=s.rstrip()
print(y)
z=s.strip()
print(z)
print(len(x))
print(len(y))
print(len(z))
```

```python
s="banana"
z=s.rstrip("na")
y=s.rstrip("a")
x=s.rstrip("an") # if we do this it removes all the an from right side to all
print(x)
print(y)
print(z)
```

```python
s="www.example.com"
x=s.strip("w.com")
print(x)
```

```python
#changing the case of string
#upper(): It changes to upper case string
s="Aditya"
print(s.upper())

#lower(): It changes to lower case string
t="Anvi"
print(t.lower())

#swapcase: It cahnges from upper to lower and vice versa
r="Abhi"
print(r.swapcase())

#title(): It gets all the first letter of word to be capital
x="Python is very easy"
print(x.title())

#capitalize(): It only prints the first letter f the line to be capital 
y="LJ UNIVERSITY"
print(x.capitalize())
print(x.title())
```

```python
# check functions to check the type of characters present in a string:
#isalnum(): Checks if it is alpha numeric. it gives true only if there are alphabets and numerics.
x="LJ123340"
print(x.isalnum())

#isalpha(): checks if it is alphabets, returns true only if has alphabets.
y="LJIET"
print(y.isalpha())

#isdigit(): checks if it is numbers, returns true only if it has numbers.
z="12345"
print(z.isdigit())

#is numeric(): It returns true for fractions, roman numbers and also check if it is a number
a="12345"
print(a.isnumeric())

#islower(): It only checks the alphabets.
b="hellow world!!!"
print(b.islower())
c=x="Hello 1234"
print(x.islower())

#istitle(): 
a="Hello And Welcome"
b="hello"
c="22 names"
d="This is %$@"
print(a.istitle())
print(b.istitle())
print(c.istitle())
print(d.istitle())
```

```python
#isidentifier():
a="MyProject"
b="Demo20"
c="2Name"
d="my demo"
print(a.isidentifier())
print(b.isidentifier())
print(c.isidentifier())
print(d.isidentifier())
#isspace():
x=" "
print(x.isspace())
```

```python
s="FSfds@#4fsf323"
alphabets=0
upper=0
lower=0
digit=0
special_symbols=0
for i in s:
    if i.isalpha()==True:
        alphabets+=1
        if i.isupper()==True:
            upper+=1
        else:
            lower+=1
    elif i.isdigit()==True:
        digit+=1
    elif i.isspace():
        pass
    else:
        special_symbols+=1
print("Alphabets:",alphabets)
print("Upper:",upper)
print("Lower:",lower)
print("Digits:",digit)
print("Special Symbols:",special_symbols)
```
