```python
#to find sum and average of upper program
s = "FSfds@#4fsf323"
sum = 0
alphabets = 0
upper = 0
lower = 0
digit = 0
special_symbols = 0
for i in s:
    if i.isalpha() == True:
        alphabets+=1
        if i.isupper() == True:
            upper+=1
        else:
            lower+=1
    elif i.isdigit() == True:
        digit+=1
        sum+=1
    elif i.isspace():
        pass
    else:
        special_symbols+=1
average = sum / digit
print("Alphabets:",alphabets)
print("Upper:",upper)
print("Lower:",lower)
print("Digits:",digit)
print("Special Symbols:",special_symbols)
print("Sum:",sum)
print("Average:",average)
```

# 9 Marks question:
**WAP to check whether the password is valid or not.** 

(1) Minimum 8 characters

(2) The alphabet must be between [a-z]

(3) Atleast one alphabet should be uppercase[A-Z]

(4) Atleast one digit between [0-9]

(5) Atleast one special symbol from(_ or @ or $)

```python
password = input("Enter Password:")
upper = 0
lower = 0
digit = 0
special_symbol = 0
digit = 0
if len(password)>=8:
    for i in password:
        if i.isupper()==True:
            upper+=1
        elif i.islower()==True:
            lower+=1
        elif i.isdigit()==True:
            digit+=1
        elif (i=='_' or i=='@'or i=='$'):
            special_symbol+=1
    if upper>0 and lower>0 and digit>0 and special_symbol>0 and (upper+lower+special_symbol+digit)==len(password):
        print("Valid Password:",password)
    else:
        print("Invalid Password")
else:
    print("Minimum characters allowed are 8")
```

# Tuple:


**Tuple is immutable. Tuple is read-only version of list. If data is fixed and nver changes then we go for tuple.**

**Insertion order is preserved. Duplicates are allowed. Heterogenous objects. We can differentiate duplicate objects by using index.**

**Tuple supports both +ve and -ve indexing. We represent tuple elements within parenthesis(()) with comma sepparator.**

**Parentheses are optional but recommended to use.**

**Example:**

t = (10,20,30,40,50)

t = (10,10.5,"Rii",[1,2,3])

t = (10,20,(30,40))

t = 10,20,30

```python
# Creating a Tuple
t = ()
print(type(t))
```

```python
t = (10) # one element gives class int if not used comma separator, if we use comma separator it will give class tuple
print(type(t))
```

```python
t = 10,20,30,40
print(type(t))
```

```python
t = tuple(range(10,20,2)) #if we want list put list instead of tuple
print(t)
```

```python
#from tuple to list
t = tuple[10,20,30]
print(type(t))
```

```python
# Accessing the elements of tuple
#(1) by using index
t = (10,20.30,40,50,60)
print(t[3])
print(t[-5])
print(t[3])
```

```python
#(2) by using slicing operation
t = (10,20,30,40,50,60)
print(t[2:5])
print(t[2:100])
print(t[:100])
print(t[::2])
print(t[5:])
print(t[-5:-1])
print(t[-5:])
print(t[-5:-1:-1])
print(t[5:1:-1])
print(t[5:1:1])
```

```python
# Mathematical operations for a tuple
# + - Concatenation
# - - Repetition

t1 = (10,20,30,40)
t2 = (50,60)
print(t1+t2)
print(t1*3)
```

```python
# Tuple Packing:
a = 10 
b = 20
c = 30
d = 40
t = a,b,c,d
print(t)
```

```python
# Tuple Unpacking:
t = (10,20,30)
a,b,c = t
print(a,b,c)
```
