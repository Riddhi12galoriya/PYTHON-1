```python
grade = input("Enter Grade: ")
city = int(input("Enter city class: "))

TA = 900
p_tax = 200

grade = grade.upper()

if grade == "A":
    basic_pay = 60000
    other_allowance = 8000

elif grade == "B":
    basic_pay = 50000
    other_allowance = 7000

elif grade == "C":
    basic_pay = 40000
    other_allowance = 6000

elif grade == "D":
    basic_pay = 30000
    other_allowance = 5000

elif grade == "E":
    basic_pay = 20000
    other_allowance = 4000

elif grade == "F":
    basic_pay = 10000
    other_allowance = 3000

else:
    print("Invalid Grade")
    exit()

# HRA according to city class
if city == 1:
    HRA = basic_pay * 0.30

elif city == 2:
    HRA = basic_pay * 0.20

elif city == 3:
    HRA = basic_pay * 0.10

else:
    print("Invalid City Class")
    exit()

# DA and Professional Tax
DA = basic_pay * 0.50
Prof_tax = basic_pay * 0.11

# Gross Pay
gp = basic_pay + HRA + DA + other_allowance + TA

# Net Pay
net_pay = gp - Prof_tax - p_tax

print("\n----- SALARY DETAILS -----")
print("Grade:", grade)
print("City Class:", city)
print("Basic Pay:", basic_pay)
print("HRA:", HRA)
print("DA:", DA)
print("Other Allowance:", other_allowance)
print("TA:", TA)
print("Gross Pay:", gp)
print("Professional Tax:", Prof_tax)
print("Professional Tax Deduction:", p_tax)
print("Net Pay:", net_pay)
```
