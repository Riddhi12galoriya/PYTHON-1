```python
#WAP using function to shift the decimal digits n places to the left, wrapping the extra digits around. If 
#shift > the number of digits of n, then reverse the string.
#Note: 
#Function will take two parameters: 
#1. The number 
#2. How much shift user want
#Example:
#Input: n=12345 shift=1
#Output: Result=23451
#Input: n=12345 shift=3
#Output: Result=45123
#Input: n=12345 shift=5
#Output: Result=12345
#Input: n=12345 shift=6
#Output: Result=54321

def shift_digits(n,shift):
    s = str(n)
    if shift > len(s):
        return s[::-1]

    else:
        shift = shift % len(s)
        result = s[shift:] + s[:shift]
        return result

n = int(input("Enter a number:"))
shift = int(input("Enter a shift value:"))
print("Result:",shift_digits(n,shift))        
```

```python
# WAP to check the validity of a Password.Primary conditions for password validation:
#1. Minimum 8 characters.
#2. The alphabet must be between [a-z]
#3. At least one alphabet should be of Upper Case [A-Z]
#4. At least 1 number or digit between [0-9]
#5. At least 1 character from [ _ or @ or $]
#Examples:
#Input: Ram@_f1234
#Output: Valid Password
#Input: Rama_fo$ab
#Output: Invalid Password
#Explanation: Number is missing
#Input: Rama#fo9c
#Output: Invalid Password
#Explanation: Must consist from _ or @ or $

password = input("Enter Password:")
upper = 0
lower = 0
digit = 0
special_symbol = 0
digit = 0
if len(password)>=8:
    for i in password:
        if i.isupper()==True:
            upper+=1
        elif i.islower()==True:
            lower+=1
        elif i.isdigit()==True:
            digit+=1
        elif (i=='_' or i=='@'or i=='$'):
            special_symbol+=1
    if upper>0 and lower>0 and digit>0 and special_symbol>0 and (upper+lower+special_symbol+digit)==len(password):
        print("Valid Password:",password)
    else:
        print("Invalid Password")
else:
    print("Minimum characters allowed are 8")
```

```python
# WAP to create a Caesar encryption.
#Note: In cryptography, a Caesar cipher, also known as Caesar's cipher, the shift cipher, Caesar's code or Caesar 
#shift, is one of the simplest and most widely known encryption techniques. It is a type of substitution cipher in 
#which each letter in the plaintext is replaced by a letter some fixed number of positions down the alphabet. For 
#example, with a right shift of 3, A would be replaced by D, E would become H, and so on. The method is named 
#after Julius Caesar, who used it in his private correspondence.
#For Example:
#Input Text  : LJIET ENG
#Shift : 3
#Cipher:  OMLHW HQJ

text = input("Enter a String:")
shift = int(input("Enter a shift value:"))
cipher = ""
for ch in text:
    if ch.isupper():
         cipher += chr((ord(ch) - ord('A') + shift) % 26 + ord('A'))
    elif ch.islower():
        cipher += chr((ord(ch) - ord('a') + shift) % 26 + ord('a'))
    else:
        cipher += ch

print("Cipher:", cipher)
```

```python
# Given a list L of size N, You need to count the number of special elements in the given list. An element is special if 
#removal of that element makes the list balanced.
#The list will be balanced if sum of even index elements is equal to the sum of odd index elements.
#Example Input
#Input 1:
#A = [2,1,6,4]
#Input 2:
#A=[5,5,2,5,8]
#Example Output
#Output 1:
#1
#Output 2:
#2
#Explanation 1 :
#After deleting 1 from list : [2,6,4]
#(2+4) = (6)
#Hence 1 is the only special element, so count is 1.
#Explanation 2 :
#If we delete A[0] or A[1], list will be balanced
#(5+5)=(2+8)
#So A[0] and A[1] are special elements, so count is 2.

def count_special(A):
    count = 0
    n = len(A)

    for i in range(n):
        B = A[:i] + A[i+1:]
        even_sum = 0
        odd_sum = 0

        for j in range(len(B)):
            if j%2==0:
                even_sum+=B[j]
            else:
                odd_sum+=B[j]
        if even_sum == odd_sum:
            count+=1
    return count
A = list(map(int, input("Enter list elements:").split()))
print("Count of special elements:",count_special(A))
```

```python
#WAP to Print Longest Common Prefix from a given list of strings. The longest common prefix for list of strings is 
#the common prefix (starting of string) between all strings.
#For example, in the given list [“apple”, “ape”, “zebra”], there is no common prefix because the 2 most dissimilar strings of the list “ape” and “zebra” do not share any starting characters.
#If there is no common prefix between all strings in the list than return -1.
#For example,
#Input list: ["lessonplan", "lesson","lees", "length"]
# The longest Common Prefix is: le
# Input list: ["python","pythonprogramming","pythonlist"]
# The longest Common Prefix is: python
# Input list: ["lessonplan", "lesson","ees", "length"]
# The longest Common Prefix is: -1

def longest_common_prefix(strings):
    prefix = strings[0]
    for s in strings[1:]:
        while not s.startswith(prefix):
            prefix = prefix[:-1]
            if prefix == "":
                return -1
    return prefix
strings = ["lessonplan", "lesson", "lees", "length"]
result = longest_common_prefix(strings)
print("The longest common prefix is:",result)
```

```python
# (B). If you get a message which is encoded by the method above then, Write a decryption program for the general case. 
# Taking input of any encrypted string from user with key number used while breaking message apart during encryption.
# For example,
# Input Encrypted message:    Hloe gl o  sogrilw g epntstfii o yotay hee nnh aoiortiimreegehrun nhnse ne
# Input Key used during encryption: 5
# Output Decrypted Message:   Hi hello how are you going to learn python in this semester of engineering
# Input Encrypted message:  Ig ntot  oopid ys lt dehaaao yrn
# Input Key used during encryption: 8
# Output Decrypted Message:  It is a good day to learn python

message = input("Input Encrypted message: ")
key = int(input("Input Key used during encryption: "))
n = len(message)
base = n // key
extra = n % key
groups = []
start = 0
for i in range(key):
    size = base + (1 if i < extra else 0)
    groups.append(message[start:start + size])
    start += size
decrypted = ""
for i in range(max(map(len, groups))):
    for group in groups:
        if i < len(group):
            decrypted += group[i]
print("Output Decrypted Message:", decrypted)
```

```python
# WAP which takes password as input and a function which checks whether the given password is valid or not under following conditions without using the RegEx module in Python language. 
# Conditions required for a valid password:
#  1.Password strength should be at least 8 characters long
#  2.Password should contain at least one uppercase and one lowercase character.
#  3.Password must have at least one number.

# Function to check password validity
def check_password(password):
    if len(password) < 8:
        return False
    upper = False
    lower = False
    digit = False
    for ch in password:
        if ch.isupper():
            upper = True
        elif ch.islower():
            lower = True
        elif ch.isdigit():
            digit = True
    if upper and lower and digit:
        return True
    else:
        return False
# Main program
password = input("Enter password: ")
if check_password(password):
    print("Valid Password")
else:
    print("Invalid Password")
```

```python
#WAP with user defined function that reads the words from paragraph and stores them as keys in a 
# dictionary and count the frequency of it as a value . 
# For Example:
# Input string: “Dog the quick brown fox jumps over the lazy dog”
# Output: {'the': 2, 'jumps': 1, 'brown': 1, 'lazy': 1, 'fox': 1, 'over': 1, 'quick': 1, 'dog': 2}

# Function to count the frequency of words
def word_frequency(sentence):
    words = sentence.lower().split()
    frequency = {}
    for word in words:
        if word in frequency:
            frequency[word] += 1
        else:
            frequency[word] = 1
    return frequency
# Main program
text = input("Input string: ")
result = word_frequency(text)
print("Output:", result)
```

```python
# Given a list L of size N. You need to count the number of special elements in the given list. An element is special if removal of 
# that element makes the list balanced. The list will be balanced if sum of even index elements is equal to the sum of odd 
# elements. Also print the updated lists after removal of special elements.
# Example 1:
# Input: 
# L=[5, 5, 2, 5, 8]
# Output:
# Original List: [5, 5, 2, 5, 8]
# Index to be removed is: 0
# List after removing index 0 : [5, 2, 5, 8]
# Original List: [5, 5, 2, 5, 8]
# Index to be removed is: 1
# List after removing index 1 : [5, 2, 5, 8]
# Total number of special elements: 2
# Explaination:
# If we delete L[0] or L[1], list will be balanced.
# [5, 2, 5, 8]
# (5+5) = (2+8)
# So L[0] and L[1] are special elements, So Count is 2.
# After removal of the special elements, list will be: [5, 2, 5, 8]
# Example 2:
# Input:
# L=[2,1,6,4]
# Output:
# Original List: [2, 1, 6, 4]
# Index to be removed is: 1
# List after removing index 1 : [2, 6, 4]
# Total Number of Special elements: 1
# Explaination:
# If we delete L[1] from list : [2,6,4]
# (2+4)  =  (6)
# Here only 1 special element. So Count is 1.
# After removal of special element, list will be : [2,6,4]

# Function to check special elements
def special_elements(L):
    count = 0
    updated_list = L.copy()
    for i in range(len(L)):
        temp = L[:i] + L[i+1:]
        even_sum = 0
        odd_sum = 0
        for j in range(len(temp)):
            if j % 2 == 0:
                even_sum += temp[j]
            else:
                odd_sum += temp[j]
        if even_sum == odd_sum:
            count += 1
            print("Original List:", L)
            print("Index to be removed is:", i)
            print("List after removing index", i, ":", temp)
            updated_list = temp
    print("Total number of special elements:", count)
    print("After removal of special elements:", updated_list)
# Main program
L = list(map(int, input("Enter list elements: ").split()))
special_elements(L)
```

```python
# WAP which will return the sum of the numbers of the list. Return 0 for an empty list. 
# Except the number 13 is very unlucky, so it does not count and number that come immediately after 13 also do not count
# in sum. 
# Example : 
# [1, 2, 3, 4] = 10 
# [] = 0
# [1, 2, 3, 4, 13] = 10 
# [13, 1, 2, 3, 13] = 5
# [1, 13, 2, 3, 4] = 8

def sum_numbers(lst):
    total = 0
    skip = False
    for num in lst:
        if skip:
            skip = False
            continue
        if num == 13:
            skip = True
            continue
        total += num
    return total
print(sum_numbers([1, 2, 3, 4]))
print(sum_numbers([]))
print(sum_numbers([1, 2, 3, 4, 13]))
print(sum_numbers([13, 1, 2, 3, 13]))
print(sum_numbers([1, 13, 2, 3, 4]))
```

```python
# WAP that accepts a list L of size N from the user and reorders its elements in the following pattern:  [L[0], 
# L[N-1], L[1], L[N-2], L[2], L[N-3], ...].
# Note:  You are not allowed to use any additional data structures like lists, strings, or dictionaries. The task must be 
# accomplished by rearranging the elements of the original list in-place, without altering the values of the elements 
# themselves.
# Example:
# Input list: [ 2, 4, 6, 8, 10]
# Output: [ 2, 10, 4, 8, 6]
# Input list: [10, 20, 30, 40]
# Output: [ 10, 40, 20, 30]

L = list(map(int, input("Enter list elements separated by space: ").split()))
left = 0
right = len(L) - 1
while left < right:
    temp = L[right]
    for i in range(right, left, -1):
        L[i] = L[i - 1]
    L[left + 1] = temp
    left += 2
    right -= 1
print("Output:", L)
```

```python
# WAP that takes a string input from the user and performs the following operations:
#  1)Print the string where each word is reversed, but the word order stays the same.
#  2)Create a dic onary with each word as a key and its length as the value.
#  3)Generate and display a list of words that contain at least 3 vowels.
#  4)Create a dic onary of character categories: vowels, consonants (le ers of the alphabet except for the vowels), digits, 
# and special characters including space, counting each type.
#  5)Display which category has the highest count.
#  6)Display the list of words with the highest number of unique le ers (case-insensi ve).
# Example:
# String:  Computers 2025 are far more powerful and efficient!!
# 1) String with each word reversed:
# sretupmoC 5202 era raf erom lufrewop dna !!tneiciffe
# 2) Dictionary with word lengths:
# {'Computers': 9, '2025': 4, 'are': 3, 'far': 3, 'more': 4, 'powerful': 8, 'and': 3, 'efficient!!': 11}
# 3) Words containing at least 3 vowels: ['Computers', 'powerful', 'efficient!!']
# 4) Character categories: {'vowels': 16, 'consonants': 23, 'digits': 4, 'special': 9}
# 5) Category with highest count: consonants
# 6) Words with the highest number of unique letters: ['computers']

s = input("Enter a string: ")
words = s.split()
# 1. Reverse each word without changing word order
print("1) String with each word reversed:")
print(" ".join(word[::-1] for word in words))

# 2. Dictionary with word lengths
lengths = {word: len(word) for word in words}
print("2) Dictionary with word lengths:")
print(lengths)

# 3. Words containing at least 3 vowels
vowels = "aeiouAEIOU"
vowel_words = []
for word in words:
    count = sum(1 for ch in word if ch in vowels)
    if count >= 3:
        vowel_words.append(word)
print("3) Words containing at least 3 vowels:")
print(vowel_words)

# 4. Count character categories
categories = {
    "vowels": 0,
    "consonants": 0,
    "digits": 0,
    "special": 0
}
for ch in s:
    if ch in vowels:
        categories["vowels"] += 1
    elif ch.isalpha():
        categories["consonants"] += 1
    elif ch.isdigit():
        categories["digits"] += 1
    else:
        categories["special"] += 1
print("4) Character categories:")
print(categories)

# 5. Category with the highest count
highest = max(categories, key=categories.get)
print("5) Category with highest count:", highest)

# 6. Words with the highest number of unique letters
unique_counts = [len(set(word.lower()) - set("0123456789!@#$%^&*()_+-=.,?")) for word in words]
max_unique = max(unique_counts, default=0)
result = [
    word.lower()
    for word, count in zip(words, unique_counts)
    if count == max_unique
]
print("6) Words with the highest number of unique letters:")
print(result)
```

```python
# Given a string S containing just the characters '(', ')', '{', '}', '[' and ']', determine if the input string is valid.
# An input string is valid if:
# 1. Open brackets must be closed by the same type of brackets.
# 2. Open brackets must be closed in the correct order.
# 3. Every close bracket has a corresponding open bracket of the same type.
# Example 1:
# Input: s = "{[()]}"
# Output: Valid
# Example 2:
# Input: s = "()[]{}"
# Output: Valid
# Example 3:
# Input: s = "[(})"
# Output: Invalid
# Example 4:
# Input: s = "({[}])"
# Output: Invalid
# Example 5:
# Input: s = "({[]]})"
# Output: Invalid

s = input("Enter a string: ")
stack = []
pairs = {')': '(', '}': '{', ']': '['}
for ch in s:
    if ch in "({[":
        stack.append(ch)
    elif ch in ")}]":
        if not stack or stack[-1] != pairs[ch]:
            print("Invalid")
            break
        stack.pop()
else:
    if len(stack) == 0:
        print("Valid")
    else:
        print("Invalid")
```
