```python
#WAP using function to shift the decimal digits n places to the left, wrapping the extra digits around. If 
#shift > the number of digits of n, then reverse the string.
#Note: 
#Function will take two parameters: 
#1. The number 
#2. How much shift user want
#Example:
#Input: n=12345 shift=1
#Output: Result=23451
#Input: n=12345 shift=3
#Output: Result=45123
#Input: n=12345 shift=5
#Output: Result=12345
#Input: n=12345 shift=6
#Output: Result=54321

def shift_digits(n,shift):
    s = str(n)
    if shift > len(s):
        return s[::-1]

    else:
        shift = shift % len(s)
        result = s[shift:] + s[:shift]
        return result

n = int(input("Enter a number:"))
shift = int(input("Enter a shift value:"))
print("Result:",shift_digits(n,shift))        
```

```python
# WAP to check the validity of a Password.Primary conditions for password validation:
#1. Minimum 8 characters.
#2. The alphabet must be between [a-z]
#3. At least one alphabet should be of Upper Case [A-Z]
#4. At least 1 number or digit between [0-9]
#5. At least 1 character from [ _ or @ or $]
#Examples:
#Input: Ram@_f1234
#Output: Valid Password
#Input: Rama_fo$ab
#Output: Invalid Password
#Explanation: Number is missing
#Input: Rama#fo9c
#Output: Invalid Password
#Explanation: Must consist from _ or @ or $

password = input("Enter Password:")
upper = 0
lower = 0
digit = 0
special_symbol = 0
digit = 0
if len(password)>=8:
    for i in password:
        if i.isupper()==True:
            upper+=1
        elif i.islower()==True:
            lower+=1
        elif i.isdigit()==True:
            digit+=1
        elif (i=='_' or i=='@'or i=='$'):
            special_symbol+=1
    if upper>0 and lower>0 and digit>0 and special_symbol>0 and (upper+lower+special_symbol+digit)==len(password):
        print("Valid Password:",password)
    else:
        print("Invalid Password")
else:
    print("Minimum characters allowed are 8")
```

```python
# WAP to create a Caesar encryption.
#Note: In cryptography, a Caesar cipher, also known as Caesar's cipher, the shift cipher, Caesar's code or Caesar 
#shift, is one of the simplest and most widely known encryption techniques. It is a type of substitution cipher in 
#which each letter in the plaintext is replaced by a letter some fixed number of positions down the alphabet. For 
#example, with a right shift of 3, A would be replaced by D, E would become H, and so on. The method is named 
#after Julius Caesar, who used it in his private correspondence.
#For Example:
#Input Text  : LJIET ENG
#Shift : 3
#Cipher:  OMLHW HQJ

text = input("Enter a String:")
shift = int(input("Enter a shift value:"))
cipher = ""
for ch in text:
    if ch.isupper():
         cipher += chr((ord(ch) - ord('A') + shift) % 26 + ord('A'))
    elif ch.islower():
        cipher += chr((ord(ch) - ord('a') + shift) % 26 + ord('a'))
    else:
        cipher += ch

print("Cipher:", cipher)
```

```python
# Given a list L of size N, You need to count the number of special elements in the given list. An element is special if 
#removal of that element makes the list balanced.
#The list will be balanced if sum of even index elements is equal to the sum of odd index elements.
#Example Input
#Input 1:
#A = [2,1,6,4]
#Input 2:
#A=[5,5,2,5,8]
#Example Output
#Output 1:
#1
#Output 2:
#2
#Explanation 1 :
#After deleting 1 from list : [2,6,4]
#(2+4) = (6)
#Hence 1 is the only special element, so count is 1.
#Explanation 2 :
#If we delete A[0] or A[1], list will be balanced
#(5+5)=(2+8)
#So A[0] and A[1] are special elements, so count is 2.

def count_special(A):
    count = 0
    n = len(A)

    for i in range(n):
        B = A[:i] + A[i+1:]
        even_sum = 0
        odd_sum = 0

        for j in range(len(B)):
            if j%2==0:
                even_sum+=B[j]
            else:
                odd_sum+=B[j]
        if even_sum == odd_sum:
            count+=1
    return count
A = list(map(int, input("Enter list elements:").split()))
print("Count of special elements:",count_special(A))
```

```python
# The Syracuse (also called Collatz or Hailstone) sequence is generated by starting with a natural number and repeatedly 
#applying the following function until reaching 1:
# syr(x) = x / 2 if x is even; and 
#syr(x) = 3x + 1 if x is odd 
#For example, the Syracuse sequence starting with 5 is: 5, 16, 8, 4, 2, 1. 
#Use of inbuilt function or math library is not allowed.
#This sequence will always go to 1 for every possible starting value.
#WAP that 
# 1.Gets a star ng value from the user 
# 2.Prints the Syracuse sequence for that star ng value

n = int(input("Enter a starting value:"))
print("Syracuse sequence:")
while n!=1:
    print(n,end=",")
    if n%2==0:
        n = n//2
    else:
        n = 3*n+1
print(1)
```
