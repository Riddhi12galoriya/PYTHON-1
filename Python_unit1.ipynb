**Logical Operators**
1. For Boolean Type Behavior:
   
(A.) and: if both arguments are true, then only result is true.

(B.) or: if one argument is true, then result is true.

(C.) not: Complement

```python
print(True and False)
print(True and True)
print(False or True)
print(not True)
print(not False)
```

2. For Non-Boolean Type Behaviour:

(A.) and: if x evaluates to false return x otherwise return y.

(B.) or: if x evaluates to true then result is x otherwise result is y

(C.) not: if x evaluates to false then result is true otherwise false.

```python
print(10 and 20) # if x=0 then it is false and returns x if x is true then it returns y
print(10 or 20) #if x is true then returns x otherwise returns y.
print(not 20)
print(not 10)
```

**Ternary Operator**

```python
a= 10
b=20
x=30 if a>b else 40
print(x)
```
