```python
n=int(input("Enter rows:"))
for i in range(1,n+1):
    for j in range(1,i+1): # spacing for front right angled triangle
        print("", end=" ")
    for k in range(i,n+1):
        print("*",end="")
    print()
```

```python
n=int(input("Enter rows:"))
for i in range(1,n+1):
    k=65
    for j in range(1,i+1):
        if i%2==0:
            print(chr(k), end=" ")
            k=k+1
        else:
            print(j, end=" ")
    print()
```

```python
n=int(input("Enter rows:"))
for i in range(1,n+1):
    for j in range(1,i+1):
        if i%2==0:
            print("#", end=" ")
        else:
            print("*", end=" ")
    print()
```

```python
n=int(input("Enter rows:"))
for i in range(1,n+1):
    for j in range(1,i+1): # spacing for front right angled triangle
        print("", end=" ")
    for k in range(i,n+1):
        print(n+1-k,end="")
    print()
```
