**if-elif-else Statement:**

    statement1

elif:

    statement2

```python
n1= int(input("Enter n1:"))
n2= int(input("Enter n2:"))
n3= int(input("Enter n3:"))
if n1>n2 and n1>n3:
    print("Biggest Number:",n1)
elif n2>n3:
    print("Biggest Number:",n2)
else:
    print("Biggest Number:",n3)
```

**if we want to create a game that can only open 7 times, we use for loop**

```python
name = ""
count = 0

while name != "adi" and count < 7:
    name = input("Enter name: ")
    count += 1

if name == "adi":
    print("Thanks for Confirmation")
else:
    print("You have used all 7 chances.")
```
