# Functions:

There are two types of functions:

**(a) Builtin Function: print(), type(), input()**

**(b) User-defined Function(UDF): myfan(), test()**

**UDF:**

Syntax:

def function_name(Parameter):  #def=mandatory
    
    body of function
    
    return value               #return=optional

```python
def wish(name):  #defining function using wish as function name
    print("Hello",name,"Good Morning") #printing the function parameters we wnat to print
wish("Avyank") #Calling the function
wish("Ekaksh")
```

# Different Categories of UDF:

**(a) Function with no parameters and no return value:**

```python
def printline():
    s=input("Enter name:")
    print(s)
printline()
```

**(b) Function with parameters and no return value:**

```python
def printline(x):
    print(x)
x=input("Enter name")
printline(x)
```

**(c) Function with parameters and return value:**

```python
def printline(s):
    return s
s=input("Enter name:")
t=printline(s)
print(t)
```

**(d) Function with no parameters and has return value**

```python
def printline():
    s=input("Enter name:")
    return s
print(printline())
```

# Types of Arguments:

1. Positional

2. Keyword

3. Default Keyboard

4. Variable lenght

```python
def f1(a,b):
    print(a+b)
f1(10,20)
#a,b=formal argument
#10,20=actual agrument
```

```python
#Keyword Argument
def wish(name,msg):
    print("hello", name, msg)
wish(name="Aditya", msg="Good morning")
wish(msg="Good morning", name="Aditya")
wish("Good morning", name="Aditya")
wish(name="Aditya", Good Morning)
```

```python
#Default Argument
def wish(name="Guest", msg="Good Morning"):
    print("Hello", name, msg)
wish("Arman")
wish()
```

```python
#Variable Length Argument:
def sum(*n):               #args=arguments  #kwargs=keyword arguments
    total=0
                            #* args is variable length argumnet
    for n1 in n:
        total=total+n1
        print("Total:",total)
sum(10)
sum(10,20,30)
sum(40,60,70,80,90)
```

```python
def f1(*s,n1):
    print(n1)
    for s1 in s:
        print(s1)
    print(n1)
f1(10)
f1(10,20,30,B=40)
f1(10,"A","B","D",50)
```

```python
def display(**n):
    print(n)
    for k,v in n.items():
        print(k,"=",v)
display(m1=100,m2=200,m3=300)
display(rno=1,name='Shivaay',marks=80,subject='python')
```

**Nested Function:**

```python
def outer():
    print("This is outer function")
    def inner():
        print("This is inner function")
    return inner()
    inner()
outer()
```
