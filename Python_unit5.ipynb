# List

We can change data in list as it is mutable.

Duplicates and heterogenous objects are allowed.

Positive, negative indexing allowed, it is dynamic.

We declare it as square brackets"[ ]". Compulsory square brackets.


```python
#Example:
l = [1,2,3,4,5]
l[1] = 20
print(l)
```

```python
#Creation of List
l = [10]
print(type(l)) # it returns output as only list when inputted in square brackets
```

```python
#Insertion of elements in list
# We can use list as typecasting also
l = list(range(1,11))
print(l)
s = "Aditya"
print(list(s))
```

```python
#eval function(dynamic input)
l = eval(input("Enter Data:"))
print(l)
print(type(l))
```

```python
#Accesing Elements of list:
# 1. By using index
l = [1,2,3,4,5,6]
print(l[1])
print(l[-2])
print(l[7])
```

```python
# 2. By using slicing operator
l = list(range(1,11))
print(l)
print(l[2:7:2])
print(l[4::2])
print(l[3:7])
print(l[8:2:-2])
print(l[4:100])
print(l[::-1])
print(l[-8:-2:-2])
print(l[-8:-2:2])
```

# List Comprehension

#### Syntax:

##### list = [expression for item in list if condition]

```python
s = [x*x for x in range(1,11)]
print(s)
```

```python
s = [x*x for x in range(1,11) if x%2==0]
print(s)
```

```python
l = list(range(1,11,2))
s = [x*x for x in l]
print(s)
```

```python
l = [22,11,35,46,78,98,65,43]
d = [x for x in l if x%2==0]
print(d)

d = [x for x in l if x%2!=0]
print(d)
```

```python
result = [x*y
         for x in range(1,5)
         for y in range(1,4)
         if (x+y)%2==0]
print(result)
```

# Dictionary:

It represents group of objects as key value pairs.

Example: rollno---name, phoneno----address.

Duplicate keys are not allowed but values can be duplicated. Heterogenous objects are allowed for both keys and values.

Insertion order is not preserved. Indexing and slicing concepts are not applicable. It is mutable.

```python
# Creating a Dictionary
d = {}
d1 = dict()
print(type(d))
print(type(d))
```

```python
l = [(10,"Aditya"),(20,"Ekaksh"),(30,"Elara")]
d = dict(l)
print(d)
```
