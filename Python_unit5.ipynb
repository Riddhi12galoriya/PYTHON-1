# List

We can change data in list as it is mutable.

Duplicates and heterogenous objects are allowed.

Positive, negative indexing allowed, it is dynamic.

We declare it as square brackets"[ ]". Compulsory square brackets.


```python
#Example:
l = [1,2,3,4,5]
l[1] = 20
print(l)
```

```python
#Creation of List
l = [10]
print(type(l)) # it returns output as only list when inputted in square brackets
```

```python
#Insertion of elements in list
# We can use list as typecasting also
l = list(range(1,11))
print(l)
s = "Aditya"
print(list(s))
```

```python
#eval function(dynamic input)
l = eval(input("Enter Data:"))
print(l)
print(type(l))
```

```python
#Accesing Elements of list:
# 1. By using index
l = [1,2,3,4,5,6]
print(l[1])
print(l[-2])
print(l[7])
```

```python
# 2. By using slicing operator
l = list(range(1,11))
print(l)
print(l[2:7:2])
print(l[4::2])
print(l[3:7])
print(l[8:2:-2])
print(l[4:100])
print(l[::-1])
print(l[-8:-2:-2])
print(l[-8:-2:2])
```

```python
#loop through list:
l = [10,20,30]
for i in l:
    print(i)
```

```python
# Mathematical operators of list:
l1 = [1,2,3]
l2 = [4,5,6]
print(l1+l2)
print(l1*3)
```

```python
c = l1+4  #wrong code
print(c)
```

```python
c = l1+[4] # Right code
print(c)
```

```python
#comparison of operator for list
x = ['Dog','Cat','Rat']
y = ['Dog','Cat','Rat']
z = ['DOG','CAT','RAT']
print(x==y)
print(x==z)
print(x!=z)
```

```python
x = [50,20,30] # it checks by elements not the length of list
y = [40,50,60,100,120]
print(x>y)
print(x>=y)
print(x<y)
print(x<=y)
```

```python
x = ['Dog','Cat','Rat']
y = ['Rat','Cat','Cat']
print(x>y)
print(x>=y)
print(x<y)
print(x<=y)
```

```python
# Membership Operator in list:
l = [1,2,3,4,5]
print(6 in l)
print(5 in l)
print(7 not in l)
```

```python
# Aliasing and cloning of list:
s = [10,20,30,40,50]
x = s.copy()
x[3] = 100
print(s)
print(id(s))
print(x)
print(id(x))
```

# Important functoies of list:

```python
#len()
l = [1,2,34,4]
print(len(l))
```

```python
#count():
l = [1,1,1,2,3,3,4,4,4,4,5]
print(l.count(1))
print(l.count(5))
print(l.count(6))
```

```python
#index():
l = [1,1,1,2,3,3,4,4,4,4,5]
print(l.index(3))
print(l.index(3,5,))
if 6 in l:
    print(l.index(6))
```

```python
#append():
l = [1,2,5,6,8]
l.append(4)
print(l)
l.append("abcd")
print(l)
l.append([1,2,3])
print(l)
```

```python
#insert():
l = [1,2,5,6,8]
l.insert(2,10)
print(l)
l.insert(2,9)
print(l)
l.insert(50,9)
print(l)
l.insert(-10,11)
print(l) # never print the functions of list direct because it returns none
l.insert(-1,13)
print(l)
```

```python
# extend():
l1 = [1,2,3]
l2 = [4,5,6]
l1.extend(l2)
print(l1)
print(l2)
print('----------------------')
l2.extend(l1)
print(l1)
print(l2)
```

```python
l1 = [1,2,3]
l1.extend("LJU")
print(l1)
```

```python
#remove(): first occurence of element.
n = [1,2,1,3,1,4,1,5]
n.remove(1)
print(n)
n.remove(1)
print(n)
n.remove(1)
print(n)
n.remove(1)
print(n)
n.remove(1)
print(n)
```

```python
n = [1,2,1,3,1,4,1,5]
x = n[:]
for i in n:
    if i==1:
        x.remove(i)
print(x)
```

```python
n = [1,2,1,3,1,4,1,5]
while 1 in n:
    n.remove(1)
print(n)
```

```python
#pop(): It removes the last element
l = [10,20,30,40,50]
print(l.pop())
print(l)
```

```python
#clear(): removes all elements in one go.
l = [10,20,30,40,50]
l.clear()
print(l)
```

```python
#reverse():reverses the list.
l = [10,20,30,40,50]
l.reverse()
print(l)
```

```python
#sorted():it sorts the elements in the list.
l = [10,30,20,50,40]
l.sort()
print(l)
```

sort() and reverse() only works in list.

```python
l = [10,30,20,50,40]
l.sort(reverse=True)
print(l)
```

```python
l = ["A","B","C","D","E","F"] #It returns the value according to ascii value
l.sort(reverse=True)
print(l)
```
