```python
d = "don't wait for your feelings to change to take action. Take the action and your feelings will change"
words = d.split()
s = {}
for word in words:
    key = word[0]
    if key in s:
        s[key].append(word)
    else:
        s[key] = [word]
print("Output:",s)
```

# Set

- It is mutable.

- If we want to represent a group of unique values as a single entity then we should go for a set.

- Duplicates are not allowed.

- Insertion order is not preserved but we can sort the elements.

- Indexing and slicing are not allowed.

- Heterogenous objects are allowed.

- We can represent the set elements within curly brackets with comma separator.

- We can apply mathematical operations like union, intersection, difference, etc on set objects.

```python
#Creation of set
s = {1,1,2,3,4,4,4,4,5,6}
print(s)
print(type(s))
```

## Mathematical functions:


```python
# 1) Union:
x = {10,20,30,40}
y = {30,40,50,60}
print(x.union(y)) # It combines or merges both the sets, if there is same elements it prints them for only one time.
print(y.union(x))
print(x|y) # We can write union like this also.
print()
print(x)
print(y)
```

```python
# 2) Intersection:
x = {10,20,30,40}
y = {30,40,50,60}
print(x.intersection(y)) # It prints common values from both.
print(x&y) # We can write intersection like this also.
```

```python
# 3) Difference:
x = {10,20,30,40}
y = {30,40,50,60}
print(x.difference(y)) # It minuses the elements and then prints it
print(x-y) # We can write difference like this also.
```

```python
# 4) Symmetric difference:
x = {10,20,30,40}
y = {30,40,50,60}
print(x.symmetric_difference(y)) # It takes the elements that are not present in both of them.
print(x^y) # We can write symmetric-difference like this also.
```

```python
# 5) issubset():
x = {'a','b','c'}
y = {'e','f','a','b','c'}
print(x.issubset(y)) # It is used when first set's elements are present in second set.
print(y.issubset(x))
```

```python
# 5) issuperset():
x = {'a','b','c'}
y = {'e','f','a','b','c'}
print(x.issuperset(y)) # It returns true if the set is not present in second set from first set.
print(y.issuperset(x))
```
